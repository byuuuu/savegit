## No 3
visualisasi data dalam bentuk 2 dimensi pakai salah satu metode Unsipervised Learning yaitu PCA, dan penjelasan datanya

```python
X = data1.drop(columns=['injuries_total'])
y = data1['injuries_total']

print("X:", X.shape)
print("y:", y.shape)
```

y yang diprediksi, X fitur

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

pca_df = pd.DataFrame(X_pca, columns=['Component A', 'Component B'])
pca_df['Label'] = y.values

pca_df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plt.scatter(pca_df['Component A'], pca_df['Component B'], s=5)
plt.xlabel("Component A")
plt.ylabel("Component B")
plt.title("Visualisasi PCA (2 Komponen)")
plt.show()
```

Berdasarkan dari visualisasi ini adalah pola sebaran data setelah direduksi yaitu pada fitur total kecelakaan yang dimana data cukup banyak dan berkumpul serta merata yang berarti kecelakaan tetap ada dan ramai sering waktu

# No 4
Menggunakan Esemble Learning dan Supervised Learning untuk Melakukan Training dan Testing


Memisahkan data latih & uji

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size=0.35, random_state=0
)

print("Train:", x_train.shape, y_train.shape)
print("Test :", x_test.shape, y_test.shape)
```

**ensemble bagging untuk regresi**

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error

rf_model = RandomForestRegressor(n_estimators=200, random_state=0)
rf_model.fit(x_train, y_train)

y_pred = rf_model.predict(x_test)

print("MSE:", mean_squared_error(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
```

**ensemble boosting untuk regresi**

```python
from sklearn.ensemble import AdaBoostRegressor
from sklearn.tree import DecisionTreeRegressor

base_model = DecisionTreeRegressor(max_depth=10, random_state=0)
adaboost_model = AdaBoostRegressor(estimator=base_model, n_estimators=200, random_state=0)

adaboost_model.fit(x_train, y_train)
y_pred = adaboost_model.predict(x_test)

print("MSE:", mean_squared_error(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
```

**gabung beberapa model**

```python
from sklearn.ensemble import StackingRegressor
from sklearn.svm import SVR
from sklearn.linear_model import LinearRegression

estimators = [
    ('rf', RandomForestRegressor(n_estimators=150, random_state=0)),
    ('svr', SVR(kernel='linear'))
]

stacking_model = StackingRegressor(
    estimators=estimators,
    final_estimator=LinearRegression()
)

stacking_model.fit(x_train, y_train)
y_pred = stacking_model.predict(x_test)

print("MSE:", mean_squared_error(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
```
